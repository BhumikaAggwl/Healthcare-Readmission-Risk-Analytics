```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind, chi2_contingency
```

```python
df = pd.read_csv("../data/processed/diabetic_cleaned.csv")
```

```python
#1. LENGTH OF STAY
# Length of stay
stay_r = df[df["Readmitted_30D"] == 1]["time_in_hospital"]
stay_nr = df[df["Readmitted_30D"] == 0]["time_in_hospital"]

t_stat_stay, p_stay = ttest_ind(stay_r, stay_nr, equal_var=False)
```

```python
# 2.NUMBER OF MEDICATIONS
# Medications
med_r = df[df["Readmitted_30D"] == 1]["num_medications"]
med_nr = df[df["Readmitted_30D"] == 0]["num_medications"]

t_stat_med, p_med = ttest_ind(med_r, med_nr, equal_var=False)
```

```python
# 3. Gender
# Gender
table = pd.crosstab(df["gender"], df["Readmitted_30D"])

chi2, p_gender, dof, expected = chi2_contingency(table)
```

```python
results = pd.DataFrame({
    "Test": [
        "Length of Stay vs Readmission",
        "Number of Medications vs Readmission",
        "Gender vs Readmission"
    ],
    "Test_Statistic": [
        t_stat_stay,
        t_stat_med,
        chi2
    ],
    "P_Value": [
        p_stay,
        p_med,
        p_gender
    ]
})

results
```

```python
results.to_csv("../data/processed/statistical_test_results.csv", index=False)
```
